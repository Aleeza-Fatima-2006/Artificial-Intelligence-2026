Loading the trained model

Loading the saved encoders

A function that'll take user's input and return the particular number saved in model

Taking user's inputs

Each categorical column is replaced with numerical value

Defined the order of features

Created a 2D numpy array

```python
model = pickle.load(open('random_forest_model.pkl', 'rb'))
encoders = joblib.load("encoders.pkl")

def encode_cols(le, value):
    value_lower = value.lower()
    for original in le.classes_:
        if original.lower() == value_lower:
            return le.transform([original])[0]
    raise ValueError(f"Invalid value '{value}'. Allowed: {list(le.classes_)}")

print("Enter user input:")
user_input = {
    "job_title": input("Job_Title: "),
    "country": input("Country: "),
    "experience_level": input("Experience level: "),
    "education_level": input("Education level: "),
    "year": int(input("Year: ")),
    "primary_skill": input("Primary skill: ")
}

for col in ['job_title', 'country', 'experience_level', 'education_level', 'primary_skill']:
    user_input[col] = encode_cols(encoders[col], user_input[col])

features = ['job_title', 'country', 'experience_level', 'education_level', 'year', 'primary_skill']
X = np.array([[user_input[col] for col in features]])

pred_salary = model.predict(X)[0]
print(f"Predicted Salary: {pred_salary:.2f}")
```
